```python
import requests



url = "https://api.coinpaprika.com/v1/tickers/{coin_id}?quotes=USD"

coins_id= {'btc-bitcoin','eth-ethereum','trx-tron'}

coin_data = []


##url.format here is used to treat placeholders in url
for coin in coins_id:
    response = requests.get(url.format(coin_id = coin))
    response.raise_for_status()
    coin_data.append(response.json())


```

```python
##The first step in transformation is picking the useful data

## The pd.json_normalized command is very useful especially in instances where you have dictionaries inside other dictionaries


##Looking at the normalized columns you can tell that the data has been flattened out into header and data instead of sub dictionaries


##this way, one can have same output for multiple coins withour feeling overwhelmed by the screen output
```

```python
## Next step is to normalize data for the three coins that we have
import pandas as pd

coins_df = pd.json_normalize(coin_data)
coins_df.head()
```

```python
## The next step is dropping the columns you dont require

coins_df.drop(columns=['first_data_at','quotes.USD.percent_change_1h', 'quotes.USD.percent_change_30m', 'quotes.USD.percent_change_6h', 
                       'quotes.USD.percent_change_12h'], inplace= True)
```

```python
## Drop a few other columns
coins_df.columns
```

```python
coins_df.drop(columns=['quotes.USD.percent_change_7d',
                       'quotes.USD.percent_change_1y',
                       'quotes.USD.percent_change_24h',
                       'quotes.USD.percent_change_30d',
                       'quotes.USD.percent_change_15m'], inplace = True)
```

```python
coins_df.head()
```

```python
coins_df.columns
```

```python
coins_df.rename(columns={'quotes.USD.price':'price_USD', 
                         'quotes.USD.volume_24h':'volume_in_24h',
                         'quotes.USD.volume_24h_change_24h':'volume_change_in_24h',
                         'quotes.USD.market_cap':'mkt_cap',
                         'quotes.USD.market_cap_change_24h':'mkt_cap_change',
                         'quotes.USD.ath_price':'ath_price', 
                         'quotes.USD.ath_date':'ath_date',
                         'quotes.USD.percent_from_price_ath':'%_from_price_ath'
                        }, inplace=True
                )
```

```python
coins_df.head()
```

```python
## Changing a string to a timestamp

coins_df['last_updated'] = pd.to_datetime(coins_df['last_updated'])
```

```python
## the final process in transformation is getting a timestamp from last updated
import datetime

coins_df['time_last_updated'] = coins_df['last_updated'].dt.time
```
